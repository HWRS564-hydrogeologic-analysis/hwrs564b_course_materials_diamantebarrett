## Adding Points to Your Map

Create point features from coordinate pairs and overlay them on your spatial data.

```python
# Define some points
# Home: 29.44962724117693, -98.72082270373754
# Stream gauge (08157540): -97.7355583333333, 30.271627777777     Waller Ck at Red River St, Austin, TX
point_list = np.array([
    [-98.720822703, 29.4496272411],   # Home
    [-97.7355583333333, 30.2716277777778]  # Stream gauge
])

# Convert to geometry
point_geom = [Point(xy) for xy in point_list]

# Create GeoDataFrame with labels
point_df = gpd.GeoDataFrame(
    {
        'Name': ['Home', 'Stream Gauge'],
        'geometry': point_geom
    },
    crs=HUC6.crs
)

# Plot
fig, ax = plt.subplots(figsize=(10, 10))

HUC6.plot(ax=ax)

# Plot each with different symbols
point_df[point_df["Name"] == "Home"].plot(
    ax=ax, color='black', marker='*', markersize=200, label='Home'
)

point_df[point_df["Name"] == "Stream Gauge"].plot(
    ax=ax, color='blue', marker='^', markersize=150, label='Stream Gauge'
)

ax.set_title("HUC Boundaries with Points of Interest")
ax.legend()

plt.show()
```

## Understanding Projections and Coordinate Reference Systems (CRS)

When working with multiple spatial datasets, it's crucial that all your data use the same Coordinate Reference System (CRS). Different datasets may come in different projections, and you need to reproject them to match before combining on a single map.

Reference: [Introduction to Coordinate Reference Systems](https://www.earthdatascience.org/courses/use-data-open-source-python/intro-vector-data-python/spatial-data-vector-shapefiles/intro-to-coordinate-reference-systems-python/)

```python
# Check the CRS of different datasets
print("HUC6 CRS:")
print(HUC6.crs)
print("\nAZ Stream Gauges CRS:")
print(gages_TX.crs)


point_df = gpd.GeoDataFrame(
    {
        'Name': ['Home', 'Stream Gauge'],
        'geometry': point_geom
    },
    crs=HUC6.crs
)

# Notice they are different! We need to reproject our points to match the gauges CRS
points_project = point_df.to_crs(gages_TX.crs)
# NOTE: .to_crs() will only work if your original spatial object has a CRS assigned 
# to it AND if that CRS is the correct CRS!

# Now plot again with reprojected points
fig, ax = plt.subplots(figsize=(10, 10))
gages_TX.plot(column='DRAIN_SQKM', categorical=False,
              legend=True, markersize=45, cmap='Set2',
              ax=ax)
# Home → star
points_project[points_project["Name"] == "Home"].plot(
    ax=ax,
    color='black',
    marker='*',
    markersize=200,
    label='Home',
    zorder=3
)

# Stream Gauge → triangle
points_project[points_project["Name"] == "Stream Gauge"].plot(
    ax=ax,
    color='blue',
    marker='^',
    markersize=150,
    label='Stream Gauge',
    zorder=3
)

ax.legend()
ax.set_title("Texas Stream Gauges with Points (Reprojected)")
plt.show()
```

## Combining Multiple Datasets on One Map

Now let's overlay multiple spatial datasets - stream gauges, points of interest, and watershed boundaries - all on the same map.

```python
# First, reproject HUC6 to match the other datasets
HUC6_project = HUC6.to_crs(gages_TX.crs)

# Now plot all three datasets together
fig, ax = plt.subplots(figsize=(10, 10))

# Plot HUC6 watershed boundaries
HUC6_project.boundary.plot(
    ax=ax,
    edgecolor='black',
    linewidth=1,
    zorder=1
)

# Plot stream gauges with drainage area colors
gages_TX.plot(
    column='DRAIN_SQKM',
    categorical=False,
    legend=True,
    markersize=25,
    cmap='Set2',
    ax=ax,
    zorder=2,
    legend_kwds={'label': "Drainage Area (km²)"}
)

# Plot Home as a star
points_project[points_project["Name"] == "Home"].plot(
    ax=ax,
    color='purple',
    marker='*',
    markersize=200,
    alpha=0.7,
    label='Home',
    zorder=3
)

# Plot Stream Gauge as a triangle
points_project[points_project["Name"] == "Stream Gauge"].plot(
    ax=ax,
    color='blue',
    marker='^',
    markersize=150,
    alpha=0.8,
    label='Selected Stream Gauge',
    zorder=3
)

ax.set_title("Texas Streams, Gauges, and Watershed Boundaries")
ax.set_xlabel("X Coordinate (m)")
ax.set_ylabel("Y Coordinate (m)")
ax.legend()

plt.show()
```

## Adding a Basemap

Use contextily to add a basemap (satellite, streets, terrain, etc.) to your spatial plots. This provides geographic context for your data.

```python
# Create the same map with a basemap background
fig, ax = plt.subplots(figsize=(10, 10))

# Plot HUC6 watershed boundaries
HUC6_project.boundary.plot(ax=ax, color=None,
                           edgecolor='black', linewidth=1,zorder=1,alpha=0.7)

# Plot stream gauges with drainage area colors
gages_TX.plot(column='DRAIN_SQKM', categorical=False,
              legend=True, markersize=25, cmap='Set2',
              ax=ax, legend_kwds={'label': "Drainage Area (km²)"})

# Plot points of interest
# Plot Home as a star
points_project[points_project["Name"] == "Home"].plot(
    ax=ax,
    color='purple',
    marker='*',
    markersize=200,
    alpha=0.7,
    label='Home',
    zorder=3
)

# Plot Stream Gauge as a triangle
points_project[points_project["Name"] == "Stream Gauge"].plot(
    ax=ax,
    color='blue',
    marker='^',
    markersize=150,
    alpha=0.8,
    label='Waller Creek Stream Gauge',
    zorder=3
)

# Add basemap - tell it to match the CRS of our data
ctx.add_basemap(ax, crs=gages_TX.crs)
ax.legend()
ax.set_title("Texas Streams, Gauges, and Watershed Boundaries")
ax.set_xlabel("Easting (m)")  
ax.set_ylabel("Northing (m)")
plt.show()
```
